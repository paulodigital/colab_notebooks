## Existe uma fórmula para calcularmos os coeficientes angular e linear da reta, que nos permitem calcular o y e ajustar esta reta.

* O coeficiente angular pode ser obtido pela seguinte equação:
formula_coeficiente.png

## Vale ressaltar que:

*   â = coeficiente angular;
*   n = número de elementos;
*   Y = Moscow;
*   x = datas.



```python
Y = Moscow
X = datas
n = np.size(Moscow)
```

```python
# Precisaremos calcular o quadrado de cada elemento de um array.
X**2
```

```python
(X**2).shape
```

```python
# Vamos ao cálculo:
(n*np.sum(X*Y) - np.sum(X)*np.sum(Y))/(n*np.sum(X**2) - np.sum(X)**2)
```

```python
# Vamos adicioná-lo à variável a, que represente o coeficiente angular:
a = (n*np.sum(X*Y) - np.sum(X)*np.sum(Y))/(n*np.sum(X**2) - np.sum(X)**2)
```

# Calculando o coeficiente linear (b)

### * Para calcular o coeficiente linear, usamos a seguinte fórmula:
coeficienteB.png


```python
b = np.mean(Y) - a*np.mean(X)
```

```python
y = a*X+b
```

```python
np.linalg.norm(Moscow-y)
```
